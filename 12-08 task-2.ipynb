```python
# Search Query Spelling Corrector

# 1. Load the corpus
with open("corpus.2.csv", "r") as f:
    vocabulary = set(
        word.strip().lower()
        for word in f
        if word.strip()
    )


# 2. Edit Distance Function
def edit_distance(a, b):
    m = len(a)
    n = len(b)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if a[i - 1] == b[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost
            )

    return dp[m][n]


# 3. Get user query
query = input("Enter search query: ")

# 4. Tokenize query
words = query.lower().split()

corrected_words = []


# 5. Check each word
for word in words:

    if word in vocabulary:
        corrected_words.append(word)
        continue

    # 6. Calculate edit distance
    distances = []

    for candidate in vocabulary:
        distance = edit_distance(word, candidate)
        distances.append((distance, candidate))

    # 7. Find closest word
    distances.sort()

    best_distance, best_word = distances[0]

    # 8. Display correction
    print("\nIncorrect word:", word)
    print("Suggested correction:", best_word)
    print("Edit distance:", best_distance)

    corrected_words.append(best_word)


# 9. Display corrected query
corrected_query = " ".join(corrected_words)

print("\nOriginal Query:", query)
print("Corrected Query:", corrected_query)
```
